```python
import pandas as pd
import matplotlib.pyplot as plt
import warnings
import numpy as np
warnings.filterwarnings("ignore")

deliveries=pd.read_csv("deliveries.csv")
matches=pd.read_csv("matches.csv")

def isbowwick(p,dk):
  if type(p)==str:
    if dk not in ["obstructing the field","run out","retired hurt"]:
      return 1
    else:
      0

df1=deliveries.copy()
df1["is_bow_wick"]=df1.apply(lambda x : isbowwick(x["player_dismissed"],x["dismissal_kind"]),axis=1)
# To know the bowler's performance against different teams
def analysis(player):
  p=df1[df1.bowler==player]
  print(p.groupby("batting_team")["is_bow_wick"].sum())
  print("total wickets:",p["is_bow_wick"].sum())
  p.groupby("batting_team")["is_bow_wick"].sum().sort_values(ascending=False).plot(kind="barh")

analysis(input(df1["bowler"].unique()))
```
