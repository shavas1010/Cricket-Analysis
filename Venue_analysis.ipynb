```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import warnings

warnings.filterwarnings("ignore")

mdf=pd.read_csv("matches.csv")
ddf=pd.read_csv("deliveries.csv")
mdf=mdf.rename(columns={"id":"match_id"})
cdf=pd.merge(mdf,ddf,on="match_id",how="left")

def inningwise(df,inning):
  df=df[df.inning==inning]
  df1=pd.DataFrame(df.groupby("venue")["total_runs"].sum()/df.groupby("venue")["match_id"].apply(lambda x : len(list(np.unique(x))))).reset_index()

  return df1
k=inningwise(cdf,1).rename(columns={0:"avg1stinn"})
h=inningwise(cdf,2).rename(columns={0:"avg2ndinn"})
j=pd.merge(k,h,on="venue")


#Toss analysis

def tosswinmatchwin(x,y):
  if x==y:
    return 1
  else:
    return 0
#win toss win match
cdf["wtwm"]=cdf.apply(lambda x : tosswinmatchwin(x["toss_winner"],x["winner"]),axis=1)
#percentage of win
wdf=cdf[cdf.wtwm==1]
wtwm=pd.DataFrame(100*wdf.groupby("venue")["match_id"].apply(lambda x: len(list(np.unique(x))))/cdf.groupby("venue")["match_id"].apply(lambda x: len(list(np.unique(x))))).reset_index()
q=pd.merge(j,wtwm,on="venue").rename(columns={"match_id":"tosswin %"})


bdf=cdf[cdf.toss_decision=="bat"]

bdf["bfwm"]=bdf.apply(lambda x : tosswinmatchwin(x["toss_winner"],x["winner"]),axis=1)
zdf=bdf[bdf.bfwm==1]
fdf=cdf[cdf.toss_decision=="field"]


fdf["ffwm"]=fdf.apply(lambda x : tosswinmatchwin(x["toss_winner"],x["winner"]),axis=1)
ldf=fdf[fdf.ffwm==1]
bfwm=pd.DataFrame(100*zdf.groupby("venue")["match_id"].apply(lambda x: len(list(np.unique(x))))/bdf.groupby("venue")["match_id"].apply(lambda x: len(list(np.unique(x))))).rename(columns={"match_id":"batfirstwinmatch%"}).reset_index()
ffwm=pd.DataFrame(100*ldf.groupby("venue")["match_id"].apply(lambda x: len(list(np.unique(x))))/fdf.groupby("venue")["match_id"].apply(lambda x: len(list(np.unique(x))))).rename(columns={"match_id":"fieldfirstwinmatch%"}).reset_index()
pd.merge(q,bfwm,on="venue").merge(ffwm,on="venue").fillna(0).sort_values(by="matches",ascending=False).reset_index(drop=True)







```
